# EDA — Sepsis Early-Warning Dataset

Exploratory analysis of the synthetic ICU cohort used to train the early-warning classifier.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/icu_stays.csv')
df.shape
```

## Class balance

```python
df['sepsis_onset_24h'].value_counts(normalize=True)
```

## Vital-sign distributions: onset vs. non-onset windows

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
cols = ['heart_rate', 'map_bp', 'resp_rate', 'lactate', 'wbc', 'spo2']
for ax, c in zip(axes.ravel(), cols):
    df.boxplot(column=c, by='sepsis_onset_24h', ax=ax)
    ax.set_title(c)
plt.tight_layout()
plt.show()
```

## Trajectory example: a patient who onsets sepsis

```python
onset_stay = df[df['sepsis_onset_24h']==1]['stay_id'].iloc[0]
traj = df[df['stay_id']==onset_stay].sort_values('window_idx')
traj.plot(x='hours_since_admission', y=['heart_rate','map_bp','lactate'], figsize=(10,4), title=f'Stay {onset_stay} vitals trajectory')
plt.show()
```
